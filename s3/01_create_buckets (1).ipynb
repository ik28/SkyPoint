# 01 - Create S3 buckets

```python
# ---- CONFIG ----
import os

# Set these in your shell before starting Jupyter:
#   export AWS_ACCESS_KEY_ID=...
#   export AWS_SECRET_ACCESS_KEY=...
AWS_ACCESS_KEY_ID = os.environ["AWS_ACCESS_KEY_ID"]
AWS_SECRET_ACCESS_KEY = os.environ["AWS_SECRET_ACCESS_KEY"]
REGION = os.getenv("AWS_REGION", "ap-southeast-7")   # opt-in region: must be enabled in your AWS account

LANDING_BUCKET = "skypoints-landing-yourname123"   # bucket names are global: change yourname123
ARCHIVE_BUCKET = "skypoints-archive-yourname123"
REJECTS_BUCKET = "skypoints-rejects-yourname123"
```

```python
import boto3
s3 = boto3.client(
    "s3",
    aws_access_key_id=AWS_ACCESS_KEY_ID,
    aws_secret_access_key=AWS_SECRET_ACCESS_KEY,
    region_name=REGION,
)
```

```python
for bucket in (LANDING_BUCKET, ARCHIVE_BUCKET, REJECTS_BUCKET):
    s3.create_bucket(Bucket=bucket, CreateBucketConfiguration={"LocationConstraint": REGION})
    print("created", bucket)
```

```python
print([b["Name"] for b in s3.list_buckets()["Buckets"]])
```
